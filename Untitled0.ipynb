```python
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

# Load the datasets
buzz_fake = pd.read_csv('/content/my_data/BuzzFeed_fake_news_content.csv')
buzz_real = pd.read_csv('/content/my_data/BuzzFeed_real_news_content.csv')
pol_fake = pd.read_csv('/content/my_data/PolitiFact_fake_news_content.csv')
pol_real = pd.read_csv('/content/my_data/PolitiFact_real_news_content.csv')


# 1. Add labels
buzz_fake['label'] = 1
buzz_real['label'] = 0
pol_fake['label'] = 1
pol_real['label'] = 0

# Combine datasets
data = pd.concat([
    buzz_fake[['title','text','label']],
    buzz_real[['title','text','label']],
    pol_fake[['title','text','label']],
    pol_real[['title','text','label']]
])

# Combine title + text
data['content'] = data['title'].fillna('') + " " + data['text'].fillna('')

# 2. Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    data['content'], data['label'], test_size=0.2, random_state=42, stratify=data['label']
)

# 3. Tokenize + pad
max_words = 20000
max_len = 300
tokenizer = Tokenizer(num_words=max_words, oov_token="<UNK>")
tokenizer.fit_on_texts(X_train)

X_train_seq = pad_sequences(tokenizer.texts_to_sequences(X_train), maxlen=max_len)
X_test_seq  = pad_sequences(tokenizer.texts_to_sequences(X_test), maxlen=max_len)

# 4. Build NN model
model = Sequential([
    Embedding(input_dim=max_words, output_dim=128, input_length=max_len),
    LSTM(128, dropout=0.2, recurrent_dropout=0.2),
    Dense(64, activation="relu"),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])

# 5. Train
history = model.fit(
    X_train_seq, y_train,
    batch_size=64,
    epochs=5,
    validation_data=(X_test_seq, y_test)
)

# Evaluate
loss, acc = model.evaluate(X_test_seq, y_test)
print(f"Test accuracy: {acc:.3f}")
```

```python
from tensorflow.keras.layers import Conv1D, GlobalMaxPooling1D

# 4. Build CNN model
cnn_model = Sequential([
    Embedding(input_dim=max_words, output_dim=128),
    Conv1D(128, 5, activation='relu'),
    GlobalMaxPooling1D(),
    Dense(64, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

cnn_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# 5. Train CNN
history_cnn = cnn_model.fit(
    X_train_seq, y_train,
    batch_size=64,
    epochs=5,
    validation_data=(X_test_seq, y_test)
)

# Evaluate CNN
loss_cnn, acc_cnn = cnn_model.evaluate(X_test_seq, y_test)
print(f"CNN Test accuracy: {acc_cnn:.3f}")
```

```python
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.metrics import F1Score
from tensorflow.keras.callbacks import EarlyStopping
import numpy as np

def create_final_model(num_layers, units_per_layer, activations, dropout_rates, learning_rate):
    """Creates a feed-forward neural network model with specified hyperparameters."""
    model = Sequential()
    model.add(Dense(units_per_layer[0], activation=activations[0], input_shape=(X_train_combined.shape[1],)))
    model.add(Dropout(dropout_rates[0]))

    for i in range(1, num_layers):
        model.add(Dense(units_per_layer[i], activation=activations[i]))
        model.add(Dropout(dropout_rates[i]))

    model.add(Dense(1, activation='sigmoid'))

    optimizer = Adam(learning_rate=learning_rate)
    model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy', F1Score(average='weighted')])
    return model

# Hyperparameter search space
param_grid = {
    'num_layers': [1, 2, 3],
    'units_per_layer': [[64], [128], [64, 32], [128, 64], [128, 64, 32]],
    'activations': [['relu'], ['relu', 'relu'], ['relu', 'relu', 'relu']],
    'dropout_rates': [[0.5], [0.6], [0.5, 0.5], [0.6, 0.6], [0.5, 0.5, 0.5], [0.6, 0.6, 0.6]],
    'learning_rate': [0.01, 0.001, 0.0001],
    'epochs': [10, 20, 30],
    'batch_size': [32, 64]
}

best_accuracy = 0
best_f1_score = 0
best_hyperparameters = {}

# Hyperparameter search loop
for num_layers in param_grid['num_layers']:
    for units_per_layer in param_grid['units_per_layer']:
        if len(units_per_layer) != num_layers:
            continue # Skip if number of units doesn't match number of layers

        for activations in param_grid['activations']:
            if len(activations) != num_layers:
                continue # Skip if number of activations doesn't match number of layers

            for dropout_rates in param_grid['dropout_rates']:
                if len(dropout_rates) != num_layers:
                    continue # Skip if number of dropout rates doesn't match number of layers

                for learning_rate in param_grid['learning_rate']:
                    for epochs in param_grid['epochs']:
                        for batch_size in param_grid['batch_size']:
                            print(f"Training with: Layers={num_layers}, Units={units_per_layer}, Activations={activations}, Dropout={dropout_rates}, LR={learning_rate}, Epochs={epochs}, Batch Size={batch_size}")

                            model = create_final_model(num_layers, units_per_layer, activations, dropout_rates, learning_rate)

                            early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

                            history = model.fit(
                                X_train_combined, y_train,
                                batch_size=batch_size,
                                epochs=epochs,
                                validation_data=(X_test_combined, y_test),
                                callbacks=[early_stopping],
                                verbose=0 # Suppress verbose output during hyperparameter search
                            )

                            loss, acc, f1 = model.evaluate(X_test_combined, y_test, verbose=0)

                            print(f"  Validation Accuracy: {acc:.3f}, Validation F1 Score: {f1:.3f}")

                            if f1 > best_f1_score:
                                best_f1_score = f1
                                best_accuracy = acc
                                best_hyperparameters = {
                                    'num_layers': num_layers,
                                    'units_per_layer': units_per_layer,
                                    'activations': activations,
                                    'dropout_rates': dropout_rates,
                                    'learning_rate': learning_rate,
                                    'epochs': epochs, # Note: actual epochs trained might be less due to early stopping
                                    'batch_size': batch_size
                                }

print("\nBest Hyperparameters:")
print(best_hyperparameters)
print(f"Best Validation Accuracy: {best_accuracy:.3f}")
print(f"Best Validation F1 Score: {best_f1_score:.3f}")
```

## Explore alternative feature combination methods

### Subtask:
Instead of simple concatenation, consider other ways to combine the features from LSTM and CNN, such as weighted averaging or more complex fusion layers.


**Reasoning**:
Define a new Keras model that takes the LSTM and CNN feature tensors as separate inputs, implement a fusion layer to combine them, add subsequent layers, compile the model, and train it.



```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, concatenate, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.metrics import F1Score

# Define the input layers for LSTM and CNN features
lstm_input = Input(shape=(lstm_train_features.shape[1],), name='lstm_input')
cnn_input = Input(shape=(cnn_train_features.shape[1],), name='cnn_input')

# Concatenate the features from both branches
# Using a Dense layer as a simple fusion layer
fused_features = concatenate([lstm_input, cnn_input])
fusion_layer = Dense(128, activation='relu')(fused_features)
fusion_layer = Dropout(0.6)(fusion_layer)


# Add subsequent Dense and Dropout layers as in the final classifier model
x = Dense(64, activation='relu')(fusion_layer)
x = Dropout(0.6)(x)

# Output layer
output = Dense(1, activation='sigmoid', name='output')(x)

# Create the fused model
fused_model = Model(inputs=[lstm_input, cnn_input], outputs=output)

# Compile the model
fused_model.compile(loss='binary_crossentropy', optimizer=Adam(learning_rate=0.001), metrics=['accuracy', F1Score(average='weighted')])

# Train the new combined model using the original X_train_seq and y_train
# The model will handle feature extraction internally, so we need to pass the original sequences
# However, since we already extracted features, we train on the extracted features.
# We need to pass the extracted features as a list corresponding to the input layers.
history_fused = fused_model.fit(
    [lstm_train_features, cnn_train_features], y_train,
    batch_size=32,
    epochs=20,
    validation_data=([lstm_test_features, cnn_test_features], y_test)
)
```

**Reasoning**:
Evaluate the performance of the new fused model on the test set using the .evaluate() method and print the results.



```python
# Evaluate the fused model
loss_fused, acc_fused, f1_fused = fused_model.evaluate([lstm_test_features, cnn_test_features], y_test)
print(f"Fused Model Test Loss: {loss_fused:.3f}")
print(f"Fused Model Test Accuracy: {acc_fused:.3f}")
print(f"Fused Model Test F1 Score: {f1_fused:.3f}")
```
